## Validation of integration routines
Verify that Monte Carlo converges at the expected $O(n^{-1/2})$ rate to the exact answer found by Gaussian quadrature of the appropriate degree on simple polynomial functions.

```python
import sys; sys.path.append('..')
import MeshFEM, mesh, integration
import numpy as np
from matplotlib import pyplot as plt
```

```python
samples = np.logspace(0, 4, 40)

def validate(m, f, deg):
    exact = integration.euclidean.gaussQuadrature(deg, m, f)
    errors = [np.abs(exact - integration.euclidean.monteCarlo(int(ns), m, f)) / np.abs(exact) for ns in samples]

    plt.loglog(samples, errors)
    plt.loglog(samples, max(errors) * pow(samples, -1/2))
    plt.xlabel('error')
    plt.xlabel('number of samples')
    plt.grid()
```

```python
def f_quadratic(ei, p):
    return p.dot(p)
def f_cubic(ei, p):
    return 1 + p[0] * p[0] * p[0] + \
               p[0] * p[0] * p[1] + \
               p[0] * p[1] * p[1] + \
               p[1] * p[1] * p[1]
def f_quartic(ei, p):
    return pow(p.dot(p), 2)
```

```python
# Planar triangle mesh
m = mesh.Mesh('../../misc/examples/meshes/square.off')
plt.figure(figsize=(16,4))
plt.subplot(1, 3, 1); validate(m, f_quadratic, 2)
plt.subplot(1, 3, 2); validate(m, f_cubic, 3)
plt.subplot(1, 3, 3); validate(m, f_quartic, 4)
```

```python
# Triangle surface mesh
m = mesh.Mesh('../../misc/examples/meshes/cube_cross.msh').boundaryMesh()
plt.figure(figsize=(16,4))
plt.subplot(1, 3, 1); validate(m, f_quadratic, 2)
plt.subplot(1, 3, 2); validate(m, f_cubic, 3)
plt.subplot(1, 3, 3); validate(m, f_quartic, 4)
```
