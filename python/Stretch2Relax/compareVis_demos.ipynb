# Line Plots Compare

```python
exp_folder = 'IterTimeData/HilbertTestOnMac_0421/'
```

```python
model_name = 'Hilbert'
method_name_list = ['RS', 'ord_newton']
# method_name_list = ['ord_newton', 'RS']
thread_num = 10
```

```python
save_fig_dir = os.path.join(exp_folder, 'Figures', model_name)
os.makedirs(save_fig_dir, exist_ok=True)
```

```python
obj_methods_list = []
time_methods_list = [] 
grad_methods_list = []
```

```python
plt.figure(figsize=(6, 12))

for method_ind, method_name in enumerate(method_name_list):
    method_data_dir = os.path.join(exp_folder, model_name, method_name, f'UVs')
    obj_arr, time_arr, grad_norm_arr, benchmark_dict = plot_video_utils.read_benchmark_data(method_data_dir, read_npz_data=False)
    obj_methods_list.append(obj_arr)
    time_methods_list.append(time_arr)
    grad_methods_list.append(grad_norm_arr)
    
    plt.subplot(2, 1, 1)
    plt.title(f'Model: {model_name}', fontsize=16)
    plt.semilogy(grad_norm_arr, label=method_name)
    plt.ylabel('Grad Norm', fontsize=14)
    plt.xlabel('Iteration', fontsize=12)
    plt.xlim(left=0)
    
    plt.subplot(2, 1, 2)
    plt.plot(time_arr, grad_norm_arr, label=method_name)
    plt.yscale('log')
    plt.ylabel('Grad Norm', fontsize=14)
    plt.xlabel('Time (s)', fontsize=12)
    plt.xlim(left=0)


plt.legend()
# plt.savefig(f'{save_fig_dir}/Hilbert_GradNormIterandTime_RSVSNewton.pdf', dpi=300, bbox_inches="tight", pad_inches=0)
```
