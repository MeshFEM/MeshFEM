# Load a UV configuration From base path

```python
def read_uv_data(directory, i):
    uv_fn = 'uv_ravel_iter_' + str(i) + '.npz'
    uv_data = np.load(os.path.join(directory, uv_fn))
    uv_ravel = uv_data['arr']
    return uv_ravel
```

```python
base_path = 'linux_exps/Stage_exps_0521_Linux/'
```

```python
option_name = 'Adaptive'
```

```python
UV_directory = os.path.join(base_path, model_name, option_name, 'UVs')
```

```python
uv_ind = 5
```

```python
uv_ravel_arr = read_uv_data(UV_directory, uv_ind)
```

```python
uv.setVars(uv_ravel_arr)
# uv.setVars(uv_init.ravel())
```
