# Metric VS Iter/Time Plots

```python
model_stats_dir = os.path.join(base_path, model_name)
```

```python
obj_grad_time_list = plot_video_utils.readConvergenceTimingData(model_stats_dir, thread_num_list, hessian_option_list)
```

```python
obj_grad_time_list[0][0].shape
```

```python
thread_ind = 0 # only 8-thread exp
num_options = len(hessian_option_list)
iterations_list = []
for i in range(num_options):
    iterations = np.arange(0, obj_grad_time_list[i][thread_ind].shape[1])
    iterations_list.append(iterations)
```

```python
color_list, line_style_list, line_width_list = plot_video_utils.getColorLineList(num_options)
```

```python
sect = None
```

```python
Metric_name = "Grad Norm"
Metric_ind = 1
```

```python
# Energy/GradNorm VS Iter/Time
plt.figure(figsize=(12, 12))

plt.subplot(2,1,1)
for i in range(num_options):
    plt_label = SolverOptionEnum.settingLabelFromSolverStr(hessian_option_list[i])
    if sect is not None:  plt.plot(iterations_list[i][:sect], obj_grad_time_list[i][thread_ind][Metric_ind][:sect], ls=line_style_list[i], lw=line_width_list[i],color=color_list[i], label=plt_label)
    else:                 plt.plot(iterations_list[i], obj_grad_time_list[i][thread_ind][Metric_ind], ls=line_style_list[i], lw=line_width_list[i], color=color_list[i], label=plt_label)
plt.title(f"Model: {model_name}", fontsize=16)
plt.yscale('log')
plt.xlabel("Iteration", fontsize=12)
plt.ylabel(Metric_name, fontsize=12)
plt.legend(fontsize=12)

# Ensure x-axis values are only positive integers
ax = plt.gca()  # Get the current axis
ax.xaxis.set_major_locator(MaxNLocator(integer=True))  # Force integer x-axis ticks
ax.set_xlim(left=0)
ax.set_ylim(bottom=0)

plt.subplot(2,1,2)
for i in range(num_options):
    plt_label = SolverOptionEnum.settingLabelFromSolverStr(hessian_option_list[i])
    if sect is not None:  plt.plot(obj_grad_time_list[i][thread_ind][2][:sect], obj_grad_time_list[i][thread_ind][Metric_ind][:sect], ls=line_style_list[i], lw=line_width_list[i],color=color_list[i], label=plt_label)
    else:                 plt.plot(obj_grad_time_list[i][thread_ind][2], obj_grad_time_list[i][thread_ind][Metric_ind], ls=line_style_list[i], lw=line_width_list[i], color=color_list[i], label=plt_label)
plt.yscale('log')
plt.xlabel("Time [sec]", fontsize=12)
plt.ylabel(Metric_name, fontsize=12)
plt.legend(fontsize=12)

ax = plt.gca()
ax.set_xlim(left=0)
ax.set_ylim(bottom=0)


plt.show()
```
