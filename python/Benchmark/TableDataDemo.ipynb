# User Extract Quickest Information across all models

```python
def extractFastestThreadExp(ModelBase_Dict, model_name, hessian_option, thread_num_list):
    model_dict = ModelBase_Dict[model_name]
    time_min = 9999999999
    time_min_thread_number = -1
    method_dict = model_dict[hessian_option]
    for thread_ind, thread_num in enumerate(thread_num_list):
        thread_dict = method_dict[thread_num]
        total_time = thread_dict['time']
        if total_time < time_min:
            time_min = total_time
            thread_min_thread_number = thread_num
    return thread_min_thread_number
```

```python
user_hessian_option_list = ['Adaptive', 'CompMajor', 'TinyAD']
```

```python
output_file_path = "ModelTXT/model_summary.txt"

with open(output_file_path, "w") as f:
    for model_ind, model_name in enumerate(model_name_list):
        for user_hessian_option in user_hessian_option_list:
            fastest_thread_num = extractFastestThreadExp(ModelBase_Dict, model_name, user_hessian_option, thread_num_list)
            model_dict = ModelBase_Dict[model_name]
            method_dict = model_dict[user_hessian_option]
            thread_dict = method_dict[fastest_thread_num]
            
            iter_num = thread_dict['iter']
            final_energy = thread_dict['final_energy']
            total_time = thread_dict['time']
            derivative_eval_time = thread_dict['hessian_eval']
            linsys_solve_time = thread_dict['linsolve']

            if user_hessian_option == 'TinyAD':
                print(
                    f'Model Name: {model_name}. Method: {user_hessian_option}. Thread Number: {fastest_thread_num}, Iter Number: {iter_num}, '
                    f'Final Energy: {final_energy:.6f}, Total Time: {total_time:.6f}, '
                    f'DerEval Time: {derivative_eval_time:.6f}, LinSolve Time: {linsys_solve_time:.6f}.',
                    f'-----------------------------------------------------------------------------',
                    file=f
                )
            else:
                numeric_fac_time = thread_dict['numeric']
                symbol_fac_time = thread_dict['symbol']
                print(
                    f'Model Name: {model_name}. Method: {user_hessian_option}. Thread Number: {fastest_thread_num}, Iter Number: {iter_num}, '
                    f'Final Energy: {final_energy:.6f}, Total Time: {total_time:.6f}, '
                    f'DerEval Time: {derivative_eval_time:.6f}, LinSolve Time: {linsys_solve_time:.6f}.',
                    file=f
                )
                print(
                    f'Model Name: {model_name}. Method: {user_hessian_option}. Thread Number: {fastest_thread_num}, '
                    f'NumFac Time: {numeric_fac_time:.6f}, SymbolFac Time: {symbol_fac_time:.6f}.',
                    f'-----------------------------------------------------------------------------',
                    file=f
                )
```
